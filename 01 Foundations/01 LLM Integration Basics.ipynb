Import Libraries

```python
import os
from google import genai
```

Load Gemini API Key

```python
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY")) 
```

Prompt

```python
prompt = """ Explain an API to a first year student. Use exactly 3 bullet points and one funny, simple analogy. Word limit : 100 words."""
```

See Available Models

```python
model_names = client.models.list()
print("Available models:", [model.name for model in model_names])
```

Call Model & Generate Response

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config={"max_output_tokens": 10000,"temperature": 0.7}
)
```

```python
print(response.text)
```
